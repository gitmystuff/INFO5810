## Where We've Been, and Why This Module Matters

### The Geometric Thread, So Far

In previous modules we have discussed that **a row of numbers, a vector, is a point in space, and understanding is figuring out that point's position, direction, or distance relative to other points.** Probability distributions are shapes. Covariance matrices are ellipses, and their eigenvectors are that ellipse's own axes. Coefficients and eigenvectors are both vectors, doing different jobs -- feature importance vs variance activity. Embeddings made the whole idea literal: a sentence *becomes* a point in hyper-dimensional space, and "meaning" becomes geometric proximity.

Every one of those methods, though, shares something worth naming explicitly: **the relationships they find are implicit.** Correlation, pattern matching, and trend discovery all share one structural trait: they tell you that something is related, using a single number, but not what kind of relationship it is, or why. A correlation of 0.87 tells you two things move together -- it doesn't tell you *why*, or *what kind* of relationship connects them. A cosine similarity score tells you two vectors point the same way -- it can't tell you one caused the other, or that they're siblings, or that one is a subclass of the other. Distance and covariance are powerful, but they're fundamentally about *strength of association*, with the meaning of that association stripped out.

**Percentiles** describe an observation's *position* relative to others — "this passenger's fare was in the 90th percentile." That's purely positional and descriptive. It says nothing about *why* that fare was high, what relationship it has to survival, or how it connects to any other fact — it's a number derived from the data's own distribution, with zero semantic content attached.

**Group comparisons** (t-tests, ANOVA) ask whether two or more groups genuinely differ, on average, more than you'd expect from random noise alone. This is a step closer to "why" than a percentile — it can tell you *that* first-class passengers survived at meaningfully different rates than third-class passengers — but it stops there. It can't tell you *because of what mechanism*: wealth, cabin location, lifeboat proximity, social norms of the era. The comparison detects a difference; it doesn't represent a reason.

**P-values** quantify something narrower still: given an assumption that there's no real effect (the null hypothesis), how surprising is this data? That's a statement about **evidence strength against chance**, not a statement about the world's structure at all. A tiny p-value tells you "this pattern probably isn't a fluke" — it says nothing about what the pattern *means*, causally or relationally.

**Here's the unifying thread across all five (correlation, pattern matching, trend discovery, percentiles, group comparisons, p-values):** every one of them answers some version of *"is there a pattern here, and how strong or surprising is it?"* — entirely within the numbers themselves, with no representation of *what kind* of relationship exists or *why* it might hold. They're all inductive: conclusions squeezed out of observed data, confidence-weighted, but silent on mechanism.

**Graphs and ontologies answer a categorically different question: *"what is this relationship, specifically, and what else follows logically from it?"*** A p-value can tell you survival and class are related beyond chance; it cannot derive, the way an ontology could, that "first-class passengers had priority lifeboat access", or connect this to unrelated knowledge about maritime safety regulations of 1912.

**None of this makes percentiles, group comparisons, or p-values less valuable** — they remain essential for exactly the job they're built for (deciding whether an observed pattern is real, and how strong it is). The distinction isn't "old tools bad, new tools good" — it's that these tools discover *whether and how strongly* something holds, while graphs and ontologies represent and reason about *what* the relationship actually is.

### From Implicit to Explicit: The Graph Universe

A graph makes a relationship **explicit** -- instead of inferring that two things are related because they happen to be close together, you draw an actual labeled connection between them: `Paris --capitalOf--> France`. That single shift unlocks something implicit methods never could: **logical inference**. If an ontology declares "every City is a Location" and "Paris is a City," a reasoner can *derive*, with certainty rather than statistical confidence, that Paris is a Location -- a new fact nobody explicitly wrote down.

This also reframes something that makes probability feel distant to a lot of people: a sample space over many variables at once is combinatorially enormous -- too large to hold in your head. But most real systems are only *sparsely* connected -- a variable typically depends directly on just a few others, not all of them. Recognizing which connections actually exist is what makes an otherwise impossible sample space tractable. That idea -- sparsity of real connections making a huge space manageable -- is the same idea underneath Bayesian Belief Networks and, as it turns out, underneath a genuinely elegant fact: invert a covariance matrix, and you get something called a **precision matrix**, where a zero entry between two variables means those two variables are conditionally independent given everything else -- in graph terms, *no edge*. Every nonzero entry *is* an edge. The covariance matrix you've been computing all semester secretly contains a graph, once you look at its inverse.

We will be talking about two different kinds of graphs:

**Graph Type 1** — a graph over variables (features/columns). This is what the precision matrix produces: nodes are things like age, fare, survived; an edge means "these two variables are statistically dependent, given everything else." A Bayesian Network is built this way too. This kind of graph answers a population-level question: "how do characteristics relate to each other, in general, across everyone?" It's the direct descendant of covariance and correlation — same question, just made explicit as edges instead of numbers.

**Graph Type 2** — a graph over observations (individual rows, specific entities). This is how ontologies work: nodes are specific things and edges are real, specific relationships between them (requiresPrerequisite, providesSafety). This is also what a Neo4j property graph is. This kind of graph answers an individual-level question: "how is this specific thing connected to that specific thing?" Two Titanic passengers being married is a Type 2 fact — it says nothing about "age" and "survival" as general concepts, it's a fact about two particular people.

### Traditional vs. Modern Knowledge Discovery

The **traditional** approach to knowledge discovery was rule-based and query-driven: structured relational databases, hand-authored expert systems, formal ontologies defined in advance by a human expert, with discovery meaning "find patterns *within* a structure someone already built." It handled tabular data well, but struggled badly with anything that didn't fit neatly into predefined categories -- free text, fuzzy relationships, meaning that resists exact keyword matching.

This course's **modern** approach discovers knowledge by learning geometric structure directly from raw, messy data -- vectors instead of hand-built rows, embeddings instead of manually authored rules. However, ontologies and logic rules belong to our older symbolic tradition -- and are proving essential *specifically because* a language model has no built-in way to know when it's wrong.

### The AI Twist

We can call this -- **neuro-symbolic AI**, the self-attention mechanism inside every transformer, a *computed graph* of connectedness: every token compared against every other token via the same dot-product similarity math, producing a weighted adjacency structure recomputed fresh for every input. An embedding space is a *learned latent space*, standing in for some true, unmeasured structure of meaning -- the same idea as a hidden factor in Factor Analysis, or a diffusion model's latent space generating an image. We'll come back to this once we've built an actual ontology.

### Warning Will Robinson: Presentism

Recall the fallacy tradition -- Monty Hall, the conjunction fallacy, the prosecutor's fallacy -- reasoning that feels obviously correct in the moment but ignores information that changes the answer. There's one more worth adding to that list, and it's directly relevant to any historical dataset (Titanic included): **presentism** -- judging people or events from the past using today's knowledge, values, or standards, rather than the context they actually lived in. A passenger's 1912 decision, judged by 2026 safety norms, will look irrational in ways that miss what was actually knowable or expected at the time. This connects directly to counterfactual reasoning (Pearl's *abduction* step specifically): properly grounding yourself in what was *actually true* at the time, before reasoning further, is the disciplined version of avoiding this trap.

### Summary

Statistical methods discover that things are related; ontologies and graphs let you declare, reason about, and share what that relationship actually is.

**Can understanding connectedness help make sample space intuitive?**

Yes, and there's a precise version of this, not just an analogy. A sample space for a *single* variable is easy — heads or tails, 1 through 6. But once you have many variables together, the *joint* sample space explodes combinatorially — every possible combination of every variable's every value. That's genuinely hard to hold in your head, and it's exactly why probability feels distant: the space is too big to intuit directly.

**"How connected we are" is the actual mathematical escape hatch from that explosion.** If every variable depended on every other variable, you'd be stuck with the full combinatorial mess. But most real systems are *sparsely* connected — a variable typically depends directly on only a few others, not all of them. **Bayesian Networks**, which we will discuss in a future module, are the formal answer to exactly this: a graph where nodes are variables and edges represent direct dependence, letting you factor an impossibly large joint distribution into a handful of small, local, manageable pieces — one per node, conditioned only on its direct connections. Understanding "how connected we are" *is* understanding which factorization of the sample space is actually tractable. This is a genuine bridge, not a loose metaphor.

**Does PCA surface features that weren't considered?**

Yes, and there's a well-known cousin of PCA built specifically around this idea: **Factor Analysis**. PCA's components are mathematically derived directions of variance; Factor Analysis goes a step further and treats those hidden combinations as **latent variables** — unmeasured underlying causes presumed to be *generating* the correlations you observe in your measured features. Titanic's PC1 (fare + pclass + parch blended) is a decent intuitive stand-in for an unmeasured "socioeconomic standing" factor — something real that plausibly exists, wasn't directly measured, and reveals itself only through its fingerprints on the features that were.

**Do eigenvectors and covariance describe the dataset as a whole, not just feature-by-feature?**

A covariance matrix is a genuinely *global* object; its eigenvectors describe the joint shape of the entire cloud, not any single column examined in isolation.

**The connection to graphs and AI:**

Take the covariance matrix and **invert it** — this produces something called the **precision matrix**. It has a remarkable, precise property: **a zero entry in the precision matrix between two variables means those two variables are conditionally independent, given every other variable in the dataset** — in graph terms, *no edge between them*. Every nonzero entry *is* an edge. Suddenly, your covariance matrix — **is a graph**, once you look at its inverse instead of itself.

Attention is a graph. A latent space is a hidden factor. A knowledge graph grounding an LLM is Wright's path diagrams and Pearson's PCA, finally shaking hands, inside the architecture powering every AI system you'll work with.


## Part 1: Core Concepts -- Ontologies, Description Logics, and OWL

* **Ontology** -- a formal, explicit specification of a shared vocabulary: what *kinds* of things exist in a domain (classes), what specific things exist (individuals), and how they relate (properties). Not a database of facts about one dataset -- a reusable *schema* for an entire domain.
* **Class** -- a category of things (`Course`, `Person`, `City`). **Individual** -- a specific instance of a class (`machine_learning`, `Paris`). **Property** -- a relationship connecting individuals (`requiresPrerequisite`, `capitalOf`), or an individual to a raw value (`hasAge`).
* **Description Logic (DL)** -- the formal mathematical logic underlying OWL. It's deliberately less expressive than full first-order logic, in exchange for a critical guarantee: DL reasoning is always *decidable* -- a reasoner is guaranteed to terminate with a definite answer, rather than potentially running forever on a genuinely undecidable question.
* **OWL (Web Ontology Language)** -- the actual W3C standard for writing ontologies, built on top of description logic. Lets you declare subclass relationships, disjoint classes (things that can never overlap), and *restrictions* -- rules like "anything satisfying this property automatically belongs to this class."
* **RDFS (RDF Schema)** -- a lighter-weight cousin of OWL, supporting basic class hierarchies and property inheritance, but without OWL's richer restriction and disjointness machinery. RDFS is the schema layer sitting on top of raw triples.
* **Reasoning / Inference** -- the actual payoff. Given an ontology's classes, properties, and restrictions, a *reasoner* (software implementing description logic's inference rules) derives new facts that were never explicitly stated -- and, just as importantly, can detect when asserted facts are logically *impossible* together.

We'll build all of this directly, using `owlready2`.

### 2.2 Build an Ontology: Course Prerequisites

Let's build a small ontology for a genuinely useful domain: courses and their prerequisites. We'll define one *restriction*-based rule: **any course that requires at least one prerequisite is automatically an `AdvancedCourse`** -- notice we never explicitly say this about any specific course; the reasoner has to figure it out.


````python
from owlready2 import *

# Note: each example in this notebook uses its own isolated World(). Without
# this, owlready2's reasoner operates on a single shared "default world" that
# silently accumulates every ontology created earlier in the session -- so a
# later, perfectly valid ontology can fail to reason simply because an EARLIER
# cell (like our deliberate inconsistency demo in 2.5) is still sitting in
# that shared space. Isolating each example avoids this entirely.

world1 = World()
onto = world1.get_ontology("http://test.org/course_onto.owl")

with onto:
    class Course(Thing): pass
    class AdvancedCourse(Course): pass

    class requiresPrerequisite(Course >> Course): pass

    # The rule: a Course that requires at least one prerequisite IS an AdvancedCourse.
    # This is a restriction, not a fact about any specific course.
    AdvancedCourse.equivalent_to = [Course & requiresPrerequisite.some(Course)]

    intro_stats = Course("intro_stats")
    linear_algebra = Course("linear_algebra")
    machine_learning = Course("machine_learning")
    machine_learning.requiresPrerequisite = [intro_stats, linear_algebra]

print("Before reasoning:")
print("  intro_stats classes:      ", [c.name for c in intro_stats.is_a])
print("  machine_learning classes: ", [c.name for c in machine_learning.is_a])
````

### 2.3 Run the Reasoner

````python
sync_reasoner([onto], infer_property_values=True)

print("After reasoning:")
print("  intro_stats classes:      ", [c.name for c in intro_stats.is_a])
print("  machine_learning classes: ", [c.name for c in machine_learning.is_a])
````

`intro_stats` has no prerequisites, so it stays a plain `Course`. `machine_learning` has two prerequisites -- the reasoner correctly derived that it's an `AdvancedCourse`, a fact nobody explicitly assigned. This is genuine logical inference, not a lookup.

### 2.4 A Second Domain: Confirming the Pattern Generalizes

One example could be a fluke of this specific setup. Let's confirm the same restriction-based pattern works in a completely different domain -- lifeboats and safety equipment.


````python
world2 = World()
onto2 = world2.get_ontology("http://test.org/safety_onto.owl")

with onto2:
    class Vehicle(Thing): pass
    class Boat(Vehicle): pass
    class Lifeboat(Boat): pass
    class SafetyEquipment(Thing): pass
    class HasSafetyRole(Thing): pass

    class providesSafety(Thing >> SafetyEquipment): pass

    # Rule: anything that provides safety equipment automatically has a safety role.
    HasSafetyRole.equivalent_to = [Thing & providesSafety.some(SafetyEquipment)]

    life_jacket = SafetyEquipment("life_jacket")
    boat1 = Lifeboat("boat1")
    boat1.providesSafety = [life_jacket]

print("Before reasoning, boat1 classes:", [c.name for c in boat1.is_a])

sync_reasoner([onto2], infer_property_values=True)

print("After reasoning, boat1 classes: ", [c.name for c in boat1.is_a])
````

Same pattern, different domain, same result: `boat1` picks up `HasSafetyRole` purely from the restriction, confirming this isn't a coincidence specific to courses.

### Line-by-Line: How This Ontology Works

**`world2 = World()`**
Creates a completely isolated workspace for this example. Recall from earlier: without this, `owlready2`'s reasoner operates on one shared space that accumulates *every* ontology built in the session -- so an unrelated broken example elsewhere could silently break this one too. `World()` prevents that entirely.

**`onto2 = world2.get_ontology("http://test.org/safety_onto.owl")`**
Creates a new, empty ontology inside that isolated world. The URL isn't fetched from the internet -- it's just a unique *name* for this ontology, the same way a variable name doesn't need to point anywhere real. Nothing exists in it yet.

**`with onto2:`**
Everything indented under this line gets added *into* `onto2`. Without it, Python wouldn't know which ontology your new classes belong to.

**`class Vehicle(Thing): pass`**
Declares a new class called `Vehicle`, which is a subclass of `Thing` -- `Thing` is the built-in root class every other class ultimately descends from (the same role `object` plays for every Python class). `pass` just means "no extra internal logic needed -- this class's *meaning* comes entirely from how it relates to other classes and individuals, not from any code inside it."

**`class Boat(Vehicle): pass`** and **`class Lifeboat(Boat): pass`**
Build a three-level hierarchy: every `Lifeboat` is a `Boat`, and every `Boat` is a `Vehicle`. This is ordinary subclassing -- nothing inferred yet, just declared directly.

**`class SafetyEquipment(Thing): pass`** and **`class HasSafetyRole(Thing): pass`**
Two more classes, unrelated to the vehicle hierarchy so far. `HasSafetyRole` is the class we'll want the reasoner to assign *automatically* later -- notice it's declared here, but nothing has been connected to it yet.

**`class providesSafety(Thing >> SafetyEquipment): pass`**
Declares a **property** (a relationship), not a class. The `>>` syntax means "this property connects something (`Thing`, i.e. anything at all) to a `SafetyEquipment`." Read `providesSafety` as a verb: *X provides-safety Y*, where Y must be a piece of safety equipment.

**`HasSafetyRole.equivalent_to = [Thing & providesSafety.some(SafetyEquipment)]`**
This is the actual *rule*, and the most important line in the cell. It tells the reasoner: **"`HasSafetyRole` means exactly the same thing as -- is logically equivalent to -- 'any Thing that provides at least one piece of SafetyEquipment.'"** `.some(...)` is description-logic notation for "there exists at least one." This is a *definition*, not a fact about any specific individual -- it applies automatically to anything that ever satisfies it, present or future.

**`life_jacket = SafetyEquipment("life_jacket")`**
Creates an actual **individual** -- a specific, named thing -- and declares it belongs to the `SafetyEquipment` class.

**`boat1 = Lifeboat("boat1")`**
Creates a second individual, `boat1`, explicitly declared as a `Lifeboat`. At this point, nothing else has been said about it.

**`boat1.providesSafety = [life_jacket]`**
Asserts the actual relationship: `boat1` provides safety via `life_jacket`. This is a plain, stated fact -- no inference involved yet.

**`print("Before reasoning, boat1 classes:", [c.name for c in boat1.is_a])`**
Checks what classes `boat1` is *currently* known to belong to. At this point, the answer is just `['Lifeboat']` -- everything so far has been explicitly declared, nothing derived.

**`sync_reasoner([onto2], infer_property_values=True)`**
This is where the actual logical work happens. The reasoner examines every class definition, every rule, and every asserted fact in `onto2`, and works out everything that must *also* be true as a logical consequence. It notices: `boat1` provides `life_jacket`, which is `SafetyEquipment` -- and `HasSafetyRole` was *defined* as exactly this condition. Therefore, `boat1` must also be a `HasSafetyRole`, even though nobody ever wrote that down directly. Passing `[onto2]` explicitly scopes the reasoning to just this one isolated ontology.

**`print("After reasoning, boat1 classes: ", [c.name for c in boat1.is_a])`**
Now the answer is `['Lifeboat', 'HasSafetyRole']` -- `HasSafetyRole` appeared with zero additional code, purely as a logical consequence of the rule and the facts already stated. This is genuine inference: a new, true fact discovered by the reasoner, not looked up or hardcoded

### 2.5 Breaking It on Purpose: Detecting a Real Inconsistency

Inference is only half the story. A reasoner's other essential job is catching **logically impossible** combinations of facts -- something no correlation or distance metric could ever do, since those methods have no concept of "impossible," only "unlikely."


````python
# This example is DELIBERATELY broken, and given its own isolated World so
# the inconsistency doesn't affect any later ontology in this notebook.
world3 = World()
onto3 = world3.get_ontology("http://test.org/animal_onto.owl")

with onto3:
    class Animal(Thing): pass
    class Herbivore(Animal): pass
    class Carnivore(Animal): pass
    AllDisjoint([Herbivore, Carnivore])   # these two classes can NEVER overlap

    rabbit = Herbivore("rabbit")

    # Deliberately create a contradiction: assert the rabbit is ALSO a carnivore
    rabbit.is_a.append(Carnivore)

print("rabbit's asserted classes:", [c.name for c in rabbit.is_a])

try:
    sync_reasoner([onto3], infer_property_values=True)
    print("Reasoner found no problem (unexpected)")
except OwlReadyInconsistentOntologyError:
    print("The reasoner correctly detected a logical INCONSISTENCY.")
    print("A rabbit cannot be both a Herbivore and a Carnivore -- these were declared disjoint.")
````

This is worth sitting with: the reasoner didn't just fail to find a pattern -- it proved, with logical certainty, that the asserted facts *cannot all be true at once*. That's a fundamentally different kind of statement than anything a p-value or a correlation coefficient can make.

### 2.6 Querying Your Ontology

Once an ontology is built and reasoned over, you can query it directly.


````python
world4 = World()
onto4 = world4.get_ontology("http://test.org/course_onto2.owl")

with onto4:
    class Course(Thing): pass
    class AdvancedCourse(Course): pass
    class requiresPrerequisite(Course >> Course): pass
    AdvancedCourse.equivalent_to = [Course & requiresPrerequisite.some(Course)]

    intro_stats = Course("intro_stats")
    linear_algebra = Course("linear_algebra")
    machine_learning = Course("machine_learning")
    deep_learning = Course("deep_learning")
    machine_learning.requiresPrerequisite = [intro_stats, linear_algebra]
    deep_learning.requiresPrerequisite = [machine_learning]   # requires an AdvancedCourse itself

sync_reasoner([onto4], infer_property_values=True)

# Query 1: every instance of a given class
print("All AdvancedCourse instances:", [c.name for c in AdvancedCourse.instances()])

# Query 2: search by a specific property value
print("\nCourses requiring intro_stats:",
      [c.name for c in onto4.search(requiresPrerequisite=intro_stats)])

# Query 3: every individual and its final, reasoned classes
print("\nAll individuals and their classes:")
for individual in onto4.individuals():
    print(f"  {individual.name}: {[c.name for c in individual.is_a]}")
````

Notice `deep_learning` also gets classified as `AdvancedCourse`, since it has its own prerequisite (`machine_learning`) -- a small cascading chain of inference, entirely automatic.

## Part 3: Tying It to AI -- Neuro-Symbolic Reasoning

**The problem:** a large language model generates the statistically most plausible next token. It has no built-in mechanism for checking whether what it just said is *true*. This is the well-known hallucination problem, and it's a direct consequence of the geometric/learned paradigm having no explicit notion of "verified fact."

**The evidence this actually works:** a 2026 clinical question-answering study found that grounding an LLM in an RDF/OWL ontology plus a knowledge graph cut hallucination from **63% down to 1.7%** , while accuracy rose from 37% to **98%** on the same domain (https://pubmed.ncbi.nlm.nih.gov/41610815/) -- the difference between a system too unreliable to deploy and one that plausibly could be. Related 2026 architectures use **Graph-Constrained Reasoning**, restricting an LLM's output so it can only produce claims traceable to real, verified paths in a knowledge graph, and enterprise systems build ontological rules directly into AI pipelines as hard compliance constraints an LLM literally cannot violate (https://arxiv.org/abs/2410.13080).

Let's build a small, working version of the core idea: checking a claim -- something an LLM might have generated -- against our verified course ontology.


````python
# world4 = World()
# onto4 = world4.get_ontology("http://test.org/course_onto2.owl")

def check_claim(course_name, claimed_class_name, ontology):
    """A mock 'AI claim verifier' -- stands in for checking an LLM's output
    against a verified ontology, without needing a live LLM API call."""
    course = ontology.search_one(iri=f"*{course_name}")
    if course is None:
        return f"UNVERIFIABLE: no course named '{course_name}' exists in the ontology."

    actual_classes = [c.name for c in course.is_a]
    if claimed_class_name in actual_classes:
        return f"VERIFIED: {course_name} is indeed classified as {claimed_class_name}."
    else:
        return (f"CONTRADICTED: the ontology classifies {course_name} as "
                f"{actual_classes}, not {claimed_class_name}.")

# Simulate three claims an LLM might generate -- one true, one false, one about
# something that doesn't exist in our verified knowledge at all.
print(check_claim("machine_learning", "AdvancedCourse", onto4))
print(check_claim("intro_stats", "AdvancedCourse", onto4))
print(check_claim("quantum_computing", "AdvancedCourse", onto4))
````

Three distinct, meaningful outcomes: a claim that checks out, a claim that's directly contradicted by verified structure, and a claim about something the ontology has no knowledge of at all. This is the actual mechanism underneath the 63%-to-1.7% hallucination reduction cited above, scaled down to something you can run and inspect directly -- an LLM's output isn't trusted at face value, it's checked against a structure that can prove it right, prove it wrong, or admit it doesn't know.


## Part 4: A Quick Contrast -- RDFS vs. Do-Calculus

Both are formal rule systems operating on a labeled graph to derive facts the graph doesn't state outright -- worth being precise about where they diverge, since you'll meet do-calculus properly in Module 12.

* **RDFS/OWL reasoning** (what we just did) is about **structure and taxonomy** -- is-a and subclass relationships, entirely symbolic. "Is this a City? Then it's a Location." No probability involved.
* **Do-calculus** (Pearl) is about **which paths in a causal graph transmit statistical dependence** -- using a graph property called d-separation to decide whether and how a causal effect can be estimated from observational data. Its three formal rules manipulate probability expressions, not class memberships.
* **The tightest way to state it:** RDFS reasons about *what kind of thing this is*; do-calculus reasons about *whether intervening here changes what you can infer about that, given this specific graph's shape*. Both use edges and graph traversal as their engine -- but RDFS's edges encode categories, while a causal DAG's edges encode causal influence, and that difference in what the edges *mean* is why the two systems' rules look so different despite both being, structurally, "graph in, new facts out."
